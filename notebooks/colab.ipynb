## Installation

```python
from pathlib import Path
import sys, zipfile, subprocess
PROJECT = Path('/content/larryconnors-trading')
if not PROJECT.exists():
    with zipfile.ZipFile('/content/larryconnors-trading.zip') as z:
        for name in z.namelist():
            assert not Path(name).is_absolute() and '..' not in Path(name).parts
        z.extractall('/content')
assert (3, 11) <= sys.version_info[:2] < (3, 14), 'Use a compatible Python runtime'
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-c', str(PROJECT/'constraints.txt'), '-e', str(PROJECT)+'[dev]'])
# Restart the runtime now if NumPy or pandas was imported before installation.
```
